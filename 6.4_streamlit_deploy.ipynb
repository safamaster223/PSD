# 6.4 Sistem Informasi Streamlit + Deploy

## Menjalankan aplikasi

```bash
pip install -r requirements.txt
streamlit run streamlit_jatim.py
```

Buka `http://localhost:8501`. Dua mode:

1. **Tebak kelas** — isi 6 nilai reflektansi band (B02, B03, B04, B08, B11, B12, skala 0–10000) → model `jatim_best.joblib` (SVM-RBF-C10, akurasi 67,7%) menghitung 6 indeks + memprediksi 1 dari 6 kelas + probabilitas.
2. **Klasifikasi CSV** — upload CSV berisi kolom B02..B12 → seluruh baris diprediksi + unduh hasil.

## Deploy

- **Web statis (dokumentasi proses + peta):** Jupyter Book ini → GitHub Pages (branch `gh-pages`, via `ghp-import`). Peta Folium interaktif: `jatim/folium_jatim_samples.html` (sebaran 2.823 sampel) dan `jatim/folium_demo_klasifikasi.html` (hasil klasifikasi + Google Satellite).
- **Aplikasi interaktif:** Streamlit Community Cloud (hubungkan repo → `streamlit_jatim.py`) atau server sendiri (`streamlit run --server.port 80`).

Peta layar-penuh (folder `maps/` di situs ini):

- [Sebaran 2.823 sampel Jatim](maps/folium_jatim_samples.html)
- [Hasil klasifikasi demo Sidoarjo–Surabaya](maps/folium_demo_klasifikasi.html)
- [Hasil klasifikasi Wonoayu sawah vs non-sawah](maps/folium_wonoayu.html)

## Batas jujur

- Akurasi 67,7%: Sawah ↔ Lahan hijau ↔ Bangunan masih tertukar (fitur cuma mean spektral). Butuh fitur tekstur/temporal untuk naik.
- Kelas Danau cuma 282 sampel (proksi Waduk).
- Peta wall-to-wall hanya jendela demo Sidoarjo–Surabaya @20 m, bukan full Jatim (full provinsi @10 m ≈ 6 GB, di luar lingkup UTS).
